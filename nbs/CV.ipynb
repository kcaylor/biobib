# CV

> Build a CV from a Numbers document.

## How to use

### Load data

Here we use the `NumbersDoc` constructor and provide the name of a numbers file.

```python
from biobib.doc import NumbersDoc
```

```python
sheet_name = "CV.numbers"
num_doc = NumbersDoc(sheet_name)
```

### Making a CV

```python
path = '../tex/CV/'
cumulative = True
```

### Service

```python
from biobib.Service import Service, service_template
```

```python
name='Service'
service_table = Service(doc=num_doc,
                        template=service_template,
                        worksheet=name,
                        table_name=name,
                        category=None,
                        cumulative=cumulative,
                        ascending=False)
```

```python
service_table.df.head()
```

### Professional Activities 

```python
from biobib.ProfessionalActivities import (
    ProfessionalActivities, 
    SpecialAppointments, 
    professional_activities_template,
    merge_tables)
```

```python
table1 = ProfessionalActivities(
    doc=num_doc,worksheet='Additional Professional Activities',
    table_name='ProfessionalActivities',
    cumulative=True,
    ascending=False,
    template=professional_activities_template
)
```

```python
table2 = SpecialAppointments(
    doc=num_doc,worksheet='Special Appointments',
    table_name='SpecialAppointments',
    cumulative=True,
    ascending=False,
    template=professional_activities_template
)
```

```python
import pandas as pd
service_table.df.rename(columns={'Committee':'Organization'}, inplace=True)
combined_activities = merge_tables(table1, table2)

# Merge the service table with the combined activities table
combined_activities.df = pd.concat([combined_activities.df, service_table.df], axis=0)

combined_activities.df['Year'] = combined_activities.df['Year'].astype('string')
# Sort the combined table by Year
combined_activities.df.sort_values(by='Year', ascending=False, inplace=True)
```

```python
combined_activities.write_template(path=path)
```

### Publications

```python
from biobib.Publications import Publications, publication_cv_template
name='Publications'
table = Publications(
    doc=num_doc,worksheet=name,
    table_name=name,
    template=publication_cv_template,
    bold_author=True,
    decorate_groups=True,
    cumulative=cumulative
)
table.write_template(path=path)
```

### Proceedings

```python
from biobib.Proceedings import Proceedings, proceedings_cv_template
table = Proceedings(
    doc=num_doc,worksheet='Conference Abstracts',
    table_name='Proceedings',
    template=proceedings_cv_template,
    bold_author=True,
    decorate_groups=False,
    cumulative=cumulative
)
table.write_template(path=path)
```
