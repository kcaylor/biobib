```python
#| export

WEB_DIR = os.getenv('WEB_DIR')
author_file = WEB_DIR + '/_data/authors.yml'
```

```python
#| export 

def normalize_name(name):
    """ Normalize names by converting them to lowercase and removing non-alphabetic characters """
    return re.sub(r'\W+', '', name.lower())
```

```python
#| export

# Assuming the YAML content is in a file named 'authors.yml'
with open(author_file, 'r') as file:
    authors = yaml.safe_load(file)

for author_name, author_details in authors.items():
    normalized_name = normalize_name(author_name)
    author_details['normalized_name'] = normalized_name

```
